# Chapter 11 executable lab demonstration
This notebook runs the same TypeScript engine used by the connected portals. All records are synthetic. No credentials or model download are required for the baseline build. The authenticated mentor journey is tested separately.

```python
from pathlib import Path
import subprocess, json
root = Path.cwd()
if root.name == "submission": root = root.parent
assert (root / "platform/package.json").exists(), "Open this notebook from the repository or submission directory"
result = subprocess.run(["npm", "run", "chapter11:build"], cwd=root / "platform", text=True, capture_output=True)
print(result.stdout)
assert result.returncode == 0, result.stderr
report = json.loads((root / "platform/artifacts/chapter11/build-acceptance.json").read_text())
```
